```python
import litellm
import os

# Replace with your read-only HF token

from google.colab import userdata
os.environ["HUGGINGFACE_API_KEY"] = userdata.get('secretName')

# Choose a reliable free model (Qwen2.5-7B is excellent for reasoning)
model_name = "huggingface/Qwen/Qwen2.5-7B-Instruct"


# Test the model
response = litellm.completion(
    model=model_name,
    messages=[{"role": "user", "content": "Say 'Hello, debate ready!' in one sentence."}],
    max_tokens=50,
    temperature=0.7
)
print(response.choices[0].message.content)
```

```python
def multi_agent_debate(question, model=model_name):
    print("🧠 Multi-Agent Debate Starting\n")
    print(f"Question: {question}\n")
    print("=" * 80)

    # Proposer
    proposer_prompt = f"""You are a confident reasoner. Answer the question with clear step-by-step reasoning, even if your first instinct might be wrong.
Question: {question}
Your answer:"""
    proposer_resp = litellm.completion(
        model=model,
        messages=[{"role": "user", "content": proposer_prompt}],
        max_tokens=512,
        temperature=0.7
    )
    proposer_answer = proposer_resp.choices[0].message.content.strip()
    print("Proposer:")
    print(proposer_answer)
    print("\n" + "—" * 80 + "\n")

    # Critic
    critic_prompt = f"""You are a sharp critic. Carefully examine the proposer's answer for any logical errors, inconsistencies, or better alternatives.
Proposer's answer: {proposer_answer}
Your critique:"""
    critic_resp = litellm.completion(
        model=model,
        messages=[{"role": "user", "content": critic_prompt}],
        max_tokens=512,
        temperature=0.7
    )
    critic_answer = critic_resp.choices[0].message.content.strip()
    print("Critic:")
    print(critic_answer)
    print("\n" + "—" * 80 + "\n")

    # Refiner
    refiner_prompt = f"""You are the final synthesizer. Combine the proposer's answer and the critic's feedback to produce the most accurate, clear final answer with full reasoning.
Proposer: {proposer_answer}
Critic: {critic_answer}
Final answer:"""
    refiner_resp = litellm.completion(
        model=model,
        messages=[{"role": "user", "content": refiner_prompt}],
        max_tokens=512,
        temperature=0.7
    )
    final_answer = refiner_resp.choices[0].message.content.strip()
    print("Refiner (Final Answer):")
    print(final_answer)

    print("\n" + "=" * 80)
    print("Debate Complete!")
    return final_answer
```

```python
question = "A bat and a ball cost $1.10 in total. The bat costs $1.00 more than the ball. How much does the ball cost? Explain your reasoning step by step."

multi_agent_debate(question)
```
